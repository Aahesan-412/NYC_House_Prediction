### Data Cleaning

```python
df.head()
```

```python
## Remove unwanted colums from df
df_clean = df.drop(columns=['id','name','host_id','host_name','last_review'])
```

```python
df_clean.head()
```

```python
## handling missing value 
df_clean['reviews_per_month'] = df_clean['reviews_per_month'].fillna(0)
```

```python
df.isnull().sum()
```

```python
df_clean.isnull().sum()
```

```python
## handling Outliers
price_cap = df_clean['price'].quantile(0.99)
nights_cap = df_clean['minimum_nights'].quantile(0.99)

df_clean['price'] = df_clean['price'].clip(upper=price_cap)
df_clean['minimum_nights'] = df_clean['minimum_nights'].clip(upper=nights_cap)
```

```python
X= df_clean.drop(columns=['room_type'])
y = df_clean['room_type']
```

### train and split

```python
from sklearn.model_selection import train_test_split

X_train , X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=42, stratify=y)
```

### Preprcessing : Columntransformer + pipeline

```python
df.columns
```

```python
numeric_cols =["longitude", "latitude","price", "minimum_nights", "number_of_reviews",
                 "reviews_per_month", "calculated_host_listings_count",
                 "availability_365"]

category_cols = ['neighbourhood_group', 'neighbourhood']
```

```python
df_clean[numeric_cols].skew()
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder , PowerTransformer
from sklearn.compose import ColumnTransformer
```

```python
## pipeline for numeric cols
numeric_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='median')),
    ('skewness', PowerTransformer(method='yeo-johnson')),
    ('scale', StandardScaler())
])

category_pipeline= Pipeline(steps=[
    ('impute', SimpleImputer(strategy='most_frequent')),
    ('encode', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ('numerical', numeric_pipeline, numeric_cols),
    ('category', category_pipeline, category_cols)
])

preprocessor
```

### Apply model

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import cross_val_score
```

```python
# class_weight="Balanced" tells the model to pay more attention to the minority class
# or rare class (Shared Room)
# but the problem is GradintBossting doesnt support class weight so we left it as it is.

models = {
    'Logistic Regression' : LogisticRegression(class_weight='balanced', random_state=42),
    'Decision Tree' : DecisionTreeClassifier(class_weight='balanced', random_state=42),
    'Random Forest' : RandomForestClassifier(class_weight='balanced', random_state=42),
    'Gradient Boosting' : GradientBoostingClassifier(random_state=42)
}

for name ,model in models.items():
    pipe = Pipeline(steps=[
        ('preprcessor', preprocessor),
        ('classifier', model)
    ])

    accuracy = cross_val_score(pipe , X_train, y_train , cv=3, scoring='accuracy')
    macrof1 = cross_val_score(pipe , X_train, y_train , cv=3, scoring='f1_macro')

    print(f"{name} -> Accuracy: {accuracy.mean():.3f} , F1: {macrof1.mean():.3f}")
```

### Hypermeter Tuning

```python
## We select the random forest from above modle
from sklearn.model_selection import RandomizedSearchCV

best_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(class_weight='balanced'))
])

param_distribution ={
    'classifier__n_estimators' : [100,150,200],
    'classifier__max_depth': [8,12,15,20,None],
    'classifier__min_samples_split': [2,5,10]
}

search = RandomizedSearchCV(
    estimator=best_pipeline,
    param_distributions=param_distribution,
    n_iter=10,
    cv=3,
    scoring='f1_macro',
    random_state=42
)

search.fit(X_train,y_train)
```

```python
print("Best Parameters: ", search.best_params_)
print("Best CV Macro-F1: ", search.best_score_)
```

### Evaluation

```python
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix
best_pipeline = search.best_estimator_

y_pred = best_pipeline.predict(X_test)

print(f"The Accuracy Score:{accuracy_score(y_test,y_pred)}")
print(f"The F1_Macro: {f1_score(y_test,y_pred, average='macro')}")
```

```python
sns.heatmap(confusion_matrix(y_test,y_pred), annot=True, fmt='.2f',
             xticklabels=best_pipeline.classes_, yticklabels=best_pipeline.classes_)
```
